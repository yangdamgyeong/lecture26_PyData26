## 1. 결측치 처리

```python
import pandas as pd
# 타이타닉 데이터 가져오기
url = 'https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'
t_df=pd.read_csv(url,sep='\t')
t_df.info()
```

```python
# 컬럼명 소문자로 변경
new_cols = ['passengerId', 'survived', 'pclass', 'name', 'sex', 'age', 'sibsp', 'parch'
            , 'ticket', 'fare', 'cabin', 'embarked']
t_df.columns=new_cols
t_df.columns
```

### 1-(1) 결측치 확인 함수

```python
# isna -> 결측치 있는 지 확인 후 있으면 NaN
t_df['age'][t_df['age'].isna()]
```

```python
# isnull -> 결측치 있으면 불 인덱스 True 반환
t_df['age'].isnull()
```

```python
# isnull -> NaN 값 추출
t_df['age'][t_df['age'].isnull()]
```

```python
# sum 함수 -> 몇개 확인
t_df['age'][t_df['age'].isnull()].isnull().sum()
```

```python
# notnull 함수 -> 결측치 없으면 True 반환
t_df['age'].notnull()
```

```python
# NaN이 아닌 값도 표시
t_df['age'][t_df['age'].notnull()]
```

```python
# loc 속성 -> age 열에서 NaN제외 데이터 검색하기
t_df.loc[t_df['age'].notnull(),['age', 'name']]
```

```python
# age 열에서 결측치인 행, name 열 값 추출 -> isnull
t_df.loc[t_df['age'].isnull(), ['age', 'name']]
```

### 1-(2) 결측치 처리 방법

```python
# 결측치를 다른 문자로 변경
t_df.head()
```

```python
# NaN 을 N으로 변경, 나머지 값은 숫자 제거 후 앞의 영문자만 남기기
t_df['cabin'] = t_df['cabin'].fillna('N')
t_df['cabin']
```

```python
# counts 함수 -> 특정 열에서 개별 값 전체 개수 확인
t_df['cabin'].value_counts()
```

```python
# 특정 열 NaN에서 N으로 변경되었는 지 확인하기
t_df['cabin'][0]
```

```python
# apply 함수와 lambda 함수 방식
t_df['cabin']=t_df['cabin'].apply(lambda x : x[0])
t_df['cabin']
```

```python
t_df.info()
```

```python
# 결측치를 평균값으로 변경하기
t_df_copy=t_df.copy()
t_df_copy
```

```python
# age열 -> 결측치 확인 True/False-> isnull
t_df_copy['age'].isnull()
```

```python
# mean 함수 -> 평균 -> fillna 함수 이용해 결측치 채우기
# 결측치인 경우만 평균값으로 변경
t_df_copy['age'] = t_df_copy['age'].fillna(t_df['age'].mean())
t_df_copy['age']
```

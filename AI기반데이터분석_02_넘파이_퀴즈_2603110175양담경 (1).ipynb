## 넘파이 퀴즈 (2026.09.01)

```python
# import문 추가
import numpy as np

arr = np.arange(12).reshape(3,4)
print(arr)
```

```python
# 배열 구성 데이터 타입
arr.dtype
```

```python
# 배열의 행과 열의 요소 갯수
arr.shape
```

```python
# 배열 요소의 전체 갯수
arr.size
```

```python
# 리스트로부터 넘파이 배열 생성
mylist = [[1, 2, 3], [4, 5, 6]]
a1 = np.array(mylist)

print(a1)
```

```python
# mylist로부터 실수형 넘파이 배열 생성
a2 = np.array(a1, dtype=float)
print(a2)
```

```python
# 3 ~ 8 숫자로 1차원 배열 생성
a3 = np.arange(3,9)

print(a3)
```

```python
# 2 ~ 15 중 짝수로 1차원 배열 생성
a4 = np.arange(2,16,2)

print(a4)
```

```python
# 0 ~ 9의 수열로  2행, 5열의 2차원 배열 생성
d1 = np.array([np.arange(5), np.arange(5,10)])

print(d1)
```

```python
d = np.arange(10)

# 배열 d의 요소를 거꾸로 출력
print(d)
print(d[::-1])
```

```python
d2 = d.reshape(2, -1)
# 1번 행 요소를 모두 50으로 변경

d2
d2[0]=50
print(d2[0])
```

```python
data = np.array([4, 6, 7, 3, 2, 0, 10, 1, 6])

# 2보다 크고 6보다 작은 요소만 추출
print(data[(data > 2) & (data < 6)])
```

```python
data = np.array([4, 6, 7, 3, 2, 0, 10, 1, 6])

# 3로 나눈 나머지가 0인 요소만 추출
print(data[data % 3 == 0])
```

```python
# 학생들의 점수가 높은 순서로 이름 출력

name_list = np.array(['John', 'Mike', 'Sarah', 'Kate', 'Samuel'])
score_list = np.array([67, 95, 84, 98, 88])

sorted_index = np.argsort(score_list)[::-1]
print(score_list[sorted_index])
print(name_list[sorted_index])
```

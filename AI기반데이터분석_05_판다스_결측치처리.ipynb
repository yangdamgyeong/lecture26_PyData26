```python
# 연습용 sample_data.csv 로딩하기
sample_file = 'data/sample_data.csv'

import pandas as pd
sample_df = pd.read_csv(sample_file)
sample_df
```

```python
sample_df.info()
```

```python
# isna
sample_df.isna()
# 컬럼 지정 가능
# sample_df['Age'].isna()
```

```python
# isnull
sample_df.isnull().sum()
```

```python
# 결측치가 있는 데이터 확인
# 'Age'컬럼에 결측치가 있는지와 결측치 데이터 확인
print(sample_df['Age'].isna().sum())
sample_df[sample_df['Age'].isna()]
```

```python
# 어떤 컬럼이든 결측치가 있는 데이터의 갯수, 데이터 확인
na_df = sample_df[sample_df.isna().any(axis=1)]
print(len(na_df))
```

```python
sample_df[sample_df.isna().any(axis=1)]
```

```python
# 결측치가 있는 데이터 삭제 : dropna
# City에 결측치가 있는 경우 데이터 삭제
sample_df.dropna(subset='City')
```

```python
sample_df = sample_df.dropna(subset='City')
sample_df
```

```python
# Age 결측치는 Age의 평균값으로 채우기
age_avg = sample_df['Age'].mean()
sample_df['Age'] = sample_df['Age'].fillna(sample_df['Age'].mean())
sample_df
```

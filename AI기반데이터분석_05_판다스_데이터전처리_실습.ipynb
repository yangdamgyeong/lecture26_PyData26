## 5-1. 그룹별로 요약하여 데이터 확인
* value_counts(), groupby(), pivot_table()

```python
sample_file = 'data/sales_sample.csv'
```

```python
# 샘플 파일 로딩
import pandas as pd
sample_df = pd.read_csv(sample_file)
sample_df.head()
```

```python
# 지역별 판매 기록의 개수
sample_df.지역.value_counts()
```

```python
# 상품별 판매 기록의 개수와 비율
sample_df.상품.value_counts()
```

```python
sample_df.상품.value_counts(normalize=True)
```

```python
# 지역과 상품의 조합별 판매 기록 개수
sample_df.value_counts(['지역', '상품'])
```

```python
# 지역별 총매출
sample_df.groupby('지역')['매출'].sum()
```

```python
# 상품별 평균 매출
sample_df.groupby('상품')['매출'].mean()
```

```python
# 지역별 판매수량의 합계와 매출의 합계
sample_df.groupby('지역')[['판매수량', '매출']].sum()
```

```python
# 지역별 매출의 건수, 합계, 평균, 최대값
sample_df.groupby('지역')['매출'].agg(['count', 'sum', 'mean', 'max'])
```

```python
# 지역과 상품별 평균 매출
sample_df.pivot_table(
    index = '지역',
    columns= '상품',
    values= '매출',
    aggfunc= 'mean'
)
```

```python
# 그룹바이로
sample_df.groupby(['지역', '상품'])['매출'].mean()
```

```python
# 지역별·상품별 총매출을 구하고 행과 열의 합계 표시
sample_df.pivot_table(
    index = '지역',
    columns= '상품',
    values= '매출',
    aggfunc= 'sum',
    margins= True
)
```

## 5-2. 결측치 처리

```python
sample_df = pd.read_csv('data/sample_data.csv')
sample_df
```

```python
# 컬럼별 결측치 갯수 확인
sample_df.isnull()
```

```python
# 컬럼별로 결측치가 있는지의 여부 확인
sample_df.isnull().any()
```

```python
# 'Age' 컬럼에 결측치가 몇개 있는지 확인
sample_df['Age'].isnull()
```

```python
sample_df['Age'].isnull().sum()
```

```python
# 'Age' 컬럼에 결측치가 있는 행 확인
sample_df[sample_df['Age'].isnull()]
```

```python
# 결측치가 하나라도 있는 행 추출 
sample_df.isnull().any(axis=1)
```

```python
sample_df.isnull().any(axis=1).sum()
```

```python
# 결측치가 하나라도 있는 행 삭제
sample_df.dropna()
```

```python
# 'City' 컬럼에 결측치가 있는 행 삭제
sample_df.dropna(subset=['City'])
```

```python
# 'Age' 컬럼의 결측치를 평균값으로 채우기]
sample_df['Age'] = sample_df['Age'].fillna(sample_df['Age'].mean())
sample_df
```

## 5-3. 중복값 처리하기

```python
# 결측치가 있어서 duplicate으로 추출되지 않을 수 있어서
# 결측치 제거 후 중복 여부 확인


# 중복값인 행 확인
sample_df.duplicated()
```

```python
sample_df.duplicated(keep=False).sum()
```

```python
# 중복값인 행 추출
sample_df[sample_df.duplicated(keep=False)]
```

```python
# 중복값인 행 삭제
sample_df.drop_duplicates()
```

```python
# 이름, 도시에 중복이 있는지 확인
sample_df[['Name', 'City']].duplicated()
```

```python
# 이름, 도시가 중복인 행 추출
duplicated_df = sample_df[sample_df.duplicated(subset=["Name", "City"], keep=False)]
```

```python
# 이름, 도시가 중복인 행 삭제
sample_df.drop_duplicates(subset=['Name', 'City'])
```

## 타이타닉 데이터 실습

```python
# 타이타닉 데이터 가져오기
titanic_data = 'https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'
titanic_df = pd.read_csv(titanic_data, sep='\t')
titanic_df.head()
```

```python
titanic_df_copy = titanic_df.copy()
```

```python
titanic_df.info()
```

```python
# 컬럼명을 소문자로 변경
titanic_df = titanic_df.rename(columns=str.lower)
```

```python
# 'age' 컬럼의 결측치 수
titanic_df.age.isnull().sum()
```

```python
# 'age' 컬럼이 결측치인 행
titanic_df[titanic_df['age'].isnull()]
```

```python
# 'age' 컬럼이 결측치인 사람들의 이름
titanic_df.loc[titanic_df['age'].isnull(), 'name']
```

```python
titanic_df.age.mean()
```

```python
# 'age'컬럼의 결측치를 age의 평균으로 채우기
titanic_df.age.fillna(titanic_df.age.mean())
```

```python
#19번 행의 데이터 확인
titanic_df.loc[19]
```

```python
titanic_df.age.isnull().sum()
```

```python
# 'cabin'의 결측치를 'N'으로 채우고, 알파벳 한자리로 변경하기
# 'cabin' 컬럼의 결측치 확인
titanic_df.cabin.isnull().sum()
```

```python
# 'cabin'의 결측치를 'N'으로 채우기
titanic_df.cabin = titanic_df.cabin.fillna('N')
titanic_df.loc[4]
```

```python
titanic_df.cabin.isnull().sum()
```

```python
# 'cabin' 컬럼의 값 확인
titanic_df.cabin.value_counts()
```

```python
# 객실의 알파벳값만 남겨 놓기
titanic_df.cabin.apply(lambda x : x[0])
```

```python
titanic_df.cabin.value_counts()
```

```python
# 'embarked' 컬럼의 결측치 삭제
titanic_df = titanic_df.dropna(subset=['embarked'])
titanic_df
```
